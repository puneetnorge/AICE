```python
import os
import random
import math
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import cv2 as cv
import shutil
from natsort import natsorted

os.environ["KMP_DUPLICATE_LIB_OK"]="TRUE"

no_of_imgs_to_use = 10 # use postive integer values 
left_indx = -math.ceil(no_of_imgs_to_use/2)+1
right_indx = math.floor(no_of_imgs_to_use/2)
print(left_indx,right_indx)
step_size = 20 # read every Nth image (sampling)
threshold_for_minimum_no_of_good_points = 50 #(80 for patient 1 and 50 for patient 2) # minimum number of good correspondence points

#use r before the path to fix unicode errors
#folder_path = r"C:\Users\psh006\OneDrive - UiT Office 365\Patients\Patient_0001\Head_1"

data_folder_names = [#r"C:\Users\psh006\OneDrive - UiT Office 365\Patients\Patient_0001\Head_1", 
                     #r"C:\Users\psh006\OneDrive - UiT Office 365\Patients\Patient_0001\Head_2",
                     r"C:\Users\psh006\OneDrive - UiT Office 365\Patients\Patient_0003\Head_1",
                     r"C:\Users\psh006\OneDrive - UiT Office 365\Patients\Patient_0003\Head_2"]
# folder path for saving corresponding images

saved_images_folder_names = [#r"C:\Users\psh006\OneDrive - UiT Office 365\ML codes\AICE\CorrespondingImages\Patient_0001\Head_1",
                             #r"C:\Users\psh006\OneDrive - UiT Office 365\ML codes\AICE\CorrespondingImages\Patient_0001\Head_2",
                             r"C:\Users\psh006\OneDrive - UiT Office 365\ML codes\AICE\CorrespondingImages\Patient_0003\Head_1",
                             r"C:\Users\psh006\OneDrive - UiT Office 365\ML codes\AICE\CorrespondingImages\Patient_0003\Head_2"]

folder_count = -1
for folder_path in data_folder_names:
    
    folder_count = folder_count + 1
    # folder path for saving corresponding images
    folder_path_saved_images = saved_images_folder_names[folder_count]
    print(folder_path_saved_images)
    
    #List all files in the folder
    file_list = os.listdir(folder_path)
    
    jpeg_files = []
    # Filter JPEG files from the file_list
    jpeg_files = [file for file in file_list if file.endswith('.jpeg') or file.endswith('.jpg')]
    
    # Sorted jpeg files 
    jpeg_files = natsorted(jpeg_files)

    # Specify the range
    start_range = 0
    end_range = len(jpeg_files)

    # Initiate SIFT detector
    sift = cv.SIFT_create()
    
    # just an image counter
    img_count = 0
    # count for sub folders names
    count = 0


    for i in range(step_size,end_range-step_size,step_size):

        img_count = img_count + 1        
        print("image no = ",img_count)       
      
        
        fname_ref = []
        fname_ref = os.path.join(folder_path, jpeg_files[i+left_indx])
        ref_fname = []
        ref_fname = jpeg_files[i+left_indx];
        
        cvImg_ref = cv.imread(fname_ref)        
        grayscale_img_ref = cv.cvtColor(cvImg_ref, cv.COLOR_BGR2GRAY)   

        kp_ref = []
        des_ref = []
        kp_ref, des_ref = sift.detectAndCompute(grayscale_img_ref,None)
        # good count when a number of images with corresponding points are identified
        good_count = 0
        
       

        for j in range(left_indx+1,right_indx+1):      

                
                
            fname_next = []
            fname_next = os.path.join(folder_path, jpeg_files[i+j])
            cvImg_next = cv.imread(fname_ref)
            grayscale_img_next = cv.cvtColor(cvImg_next, cv.COLOR_BGR2GRAY)   

            kp_next = []
            des_next = []
            kp_next, des_next = sift.detectAndCompute(grayscale_img_next,None)

            # BFMatcher with default params ()
            bf = cv.BFMatcher()
            # finding matches from BFMatcher()
            matches = bf.knnMatch(des_ref, des_next, k=2) 
            print("Total no of matches: ",len(matches))
            # select a set of good matches based on David Lowe’s ratio test
            good_matches = []

            for m1, m2 in matches:
                  if m1.distance < 0.9*m2.distance:
                        good_matches.append([m1])

            print("Number of good matches based on David Lowe’s ratio test: ",len(good_matches))

            if len(good_matches)>= threshold_for_minimum_no_of_good_points: 
                good_count = good_count + 1
            else:
                print("Not enough good corresponding points found")
        # If 10 consective images are similar enough 
        if good_count == no_of_imgs_to_use-1:
            count = count + 1
            # Specify the folder path
            new_folder_path = os.path.join(folder_path_saved_images+"\\" + str(count)+"\\")
            # Create the new folders recursively
            os.makedirs(os.path.dirname(new_folder_path), exist_ok=True)
            for jj in range(-1*good_count,1):
                file = os.path.join(folder_path, jpeg_files[i+jj])
                shutil.copy(file, new_folder_path)
                print(file)
            
            
                
    

                
                
                 
       

```
